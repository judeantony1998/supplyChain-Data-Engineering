```python
from pyspark.sql.functions import current_timestamp, col, lit

CATALOG = "capstone_supply_chain"

# ---- Purchase orders ----
df_po = (spark.read
    .option("header", "true")        # first row = column names
    .option("inferSchema", "true")   # let Spark guess number/text types
    .csv(f"/Volumes/{CATALOG}/bronze/landing_csv/purchase_orders.csv")
    .withColumn("_ingested_at", current_timestamp())         # when we loaded it
    .withColumn("_source_file", col("_metadata.file_path"))  # which file it came from
    .withColumn("_file_format", lit("csv"))                  # what format it was
)

df_po.write.mode("overwrite").saveAsTable(f"{CATALOG}.bronze.raw_purchase_orders")
print("✅ raw_purchase_orders rows:", df_po.count())
display(df_po.limit(5))

# ---- Delivery schedule (intentionally messy) ----
df_del = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(f"/Volumes/{CATALOG}/bronze/landing_csv/delivery_schedule.csv")
    .withColumn("_ingested_at", current_timestamp())
    .withColumn("_source_file", col("_metadata.file_path"))
    .withColumn("_file_format", lit("csv"))
)

df_del.write.mode("overwrite").saveAsTable(f"{CATALOG}.bronze.raw_delivery_schedule")
print("✅ raw_delivery_schedule rows:", df_del.count())
display(df_del)
```
