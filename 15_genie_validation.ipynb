```python
spark.sql("""
    SELECT vendor_name, country, quality_grade, composite_score
    FROM capstone_supply_chain.gold.agg_vendor_scorecard
    ORDER BY composite_score DESC
    LIMIT 5
""").show(truncate=False)
```

```python
spark.sql("""
    SELECT vendor_name, country, quality_grade, defect_rate_ppm, on_time_delivery_pct, composite_score
    FROM capstone_supply_chain.gold.agg_vendor_scorecard
    WHERE risk_rating = 'High'
""").show(truncate=False)

spark.sql("""
    SELECT po_id, vendor_name, part_number, order_qty, line_total, currency,
           order_date, expected_delivery, plant_name, status
    FROM capstone_supply_chain.gold.fact_purchase_orders
    WHERE status = 'Quality Hold'
""").show(truncate=False)
```
