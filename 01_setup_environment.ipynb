```python
# ENVIRONMENT SETUP
# Run this cell to create the Unity Catalog structure
# ============================================================
# NOTE: You need CREATE CATALOG permissions. If not, ask your admin
# or use an existing catalog and adjust the names below.

CATALOG = "capstone_supply_chain"

# Create catalog (skip if using existing)
spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG}")
spark.sql(f"USE CATALOG {CATALOG}")

# Create schemas for medallion architecture
for schema in ["bronze", "silver", "gold", "silver_aggregated", "gold_consolidated"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")
    print(f"Created schema: {CATALOG}.{schema}")

# Create Volumes for file landing zones
for vol_name in ["landing_csv", "landing_tsv", "landing_excel", "landing_json", "landing_pdf", "landing_images", "landing_unstructured"]:
    spark.sql(f"CREATE VOLUME IF NOT EXISTS bronze.{vol_name}")
    print(f"Created volume: {CATALOG}.bronze.{vol_name}")

print("\n✅ Environment setup complete!")
```
