```python
#Load the TSV file into Bronze

from pyspark.sql.functions import current_timestamp, col, lit

CATALOG = "capstone_supply_chain"

df_vendor = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .option("sep", "\t")   # the only difference from CSV: tab separator
    .csv(f"/Volumes/{CATALOG}/bronze/landing_tsv/vendor_performance.tsv")
    .withColumn("_ingested_at", current_timestamp())
    .withColumn("_source_file", col("_metadata.file_path"))   # replaces input_file_name()
    .withColumn("_file_format", lit("tsv"))
)

df_vendor.write.mode("overwrite").saveAsTable(f"{CATALOG}.bronze.raw_vendor_performance")
print("✅ raw_vendor_performance rows:", df_vendor.count())
display(df_vendor.limit(5))
```
